```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

output_dir = "./output"
fold_raw_data = {}
sims = {}

for i in range(1, 6):
    csv_path = os.path.join(output_dir, f"test_eval_fold_{i}_recovery.csv")
    data = pd.read_csv(csv_path)
    
    recovery_dict = dict(zip(data['Struct_ID'], data['Recovery']))
    sims[i] = recovery_dict

    fold_raw_data[i] = list(recovery_dict.values())

fold_stats = {}
for fold_num, recovery_dict in sims.items():
    recovery_vals = list(recovery_dict.values())
    mean_val = np.mean(recovery_vals)
    std_val = np.std(recovery_vals, ddof=1)
    fold_stats[fold_num] = {'mean': mean_val, 'std': std_val}

fold_nums = sorted(fold_stats.keys())
mean_values = [fold_stats[num]['mean'] for num in fold_nums]
std_values = [fold_stats[num]['std'] for num in fold_nums]

violin_box_data = [fold_raw_data[num] for num in fold_nums]

print("Statistical results of maximum Recovery for each fold:")
for fold, stats in fold_stats.items():
    print(f"Fold {fold}: Mean={stats['mean']:.4f}, Std={stats['std']:.4f}")
overall_mean = np.mean(mean_values)
print(f"\nOverall 5-Fold Mean: {overall_mean:.4f}")

fig, ax = plt.subplots(figsize=(6, 4.5), dpi=300)

violin_parts = ax.violinplot(
    violin_box_data,
    positions=fold_nums,
    showmeans=False,
    showmedians=False,
    showextrema=False,
    widths=0.6, 
    points=200
)

for pc in violin_parts['bodies']:
    pc.set_facecolor('#E0EDFE')
    pc.set_edgecolor('#3E63AD')
    pc.set_alpha(1)
    pc.set_linewidth(1.2)

box_parts = ax.boxplot(
    violin_box_data,
    positions=fold_nums,
    widths=0.2,
    patch_artist=True,
    showmeans=True,
    meanprops={
        'marker': 'o', 'markerfacecolor': 'white', 'markeredgecolor': '#3E63AD',
        'markeredgewidth': 0.8, 'markersize': 3, 'zorder': 5 
    },
    medianprops={'color': 'white', 'linewidth': 0.8, 'zorder': 4},
    boxprops={'facecolor': 'none', 'edgecolor': 'white', 'linewidth': 0.8, 'zorder': 4},
    whiskerprops={'color': 'white', 'linewidth': 0.8, 'zorder': 4},
    capprops={'color': 'white', 'linewidth': 0.8, 'zorder': 4},
    flierprops={'marker': 'x', 'markeredgecolor': 'gray', 'markersize': 4, 'zorder': 4, 'linewidth': 0.6}
)


ax.set_xlabel('5-Fold Cross Validation', fontsize=15)
ax.set_ylabel('Recovery Rate', fontsize=15)

ax.set_xticks(fold_nums)
ax.tick_params(axis='both', direction='in', length=3, width=1.2, labelsize=12)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

for spine in ['left', 'bottom']:
    ax.spines[spine].set_linewidth(0.78)

plt.tight_layout()

fig_png = os.path.join(output_dir, '5fold_recovery_violin_train_box.png')
fig_pdf = os.path.join(output_dir, '5fold_recovery_violin_train_box.pdf')
fig_svg = os.path.join(output_dir, '5fold_recovery_violin_train_box.svg')

# plt.savefig(fig_png, dpi=300, bbox_inches='tight')
# plt.savefig(fig_pdf, dpi=600, bbox_inches='tight')
# plt.savefig(fig_svg, format='svg', bbox_inches='tight')

plt.show()
```
